```python
import numpy as np
```

```python
#1. Arrays
#A NumPy array is like a list, but all items share one type and it supports fast math on the whole thing at once.
arr = np.array([1,2,3,4,5])
print(arr)
print(type(arr))
```

```python
#difference between list and array
#list 
list1 = [1,2,3]
print(list1*2)

#array 
arr1 = np.array([1,2,3])
print(arr1*2)

#differnece is that in list it will repeat the list but in array it will multiply each element by 2
```
